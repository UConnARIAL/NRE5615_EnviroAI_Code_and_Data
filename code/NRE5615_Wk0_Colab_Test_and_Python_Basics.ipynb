## 2. Variables

A variable stores a value. In environmental examples, variables might store temperature, rainfall, streamflow, or air-quality values.

```python
temperature_c = 24.5
rainfall_mm = 12
site_name = 'River monitoring station A'

print('Site:', site_name)
print('Temperature (C):', temperature_c)
print('Rainfall (mm):', rainfall_mm)
```

## 3. Simple calculations

Python can be used like a calculator. Below, we convert temperature from Celsius to Fahrenheit.

```python
temperature_f = (temperature_c * 9/5) + 32
print('Temperature in Fahrenheit:', temperature_f)
```

## 4. Lists of values

Environmental datasets often contain many measurements. A list can store multiple values.

```python
daily_temps = [21.2, 22.5, 24.1, 23.8, 25.0, 26.3, 24.9]

print('Daily temperatures:', daily_temps)
print('Number of observations:', len(daily_temps))
print('First value:', daily_temps[0])
```

## 5. Summary statistics

We often start by calculating simple summaries, such as the average, minimum, and maximum.

```python
average_temp = sum(daily_temps) / len(daily_temps)
minimum_temp = min(daily_temps)
maximum_temp = max(daily_temps)

print('Average temperature:', round(average_temp, 2))
print('Minimum temperature:', minimum_temp)
print('Maximum temperature:', maximum_temp)
```

## 6. Conditions: checking for a threshold

A condition allows Python to make a simple decision. For example, we can flag a high-temperature day.

```python
today_temp = 31.5

if today_temp >= 30:
    print('High temperature alert: review possible heat risk.')
else:
    print('Temperature is below the alert threshold.')
```

## 7. Loops: repeat an action

A loop repeats the same action for each item in a list. Here, we check each daily temperature against a simple threshold.

```python
for temp in daily_temps:
    if temp >= 25:
        print(temp, 'C: above threshold')
    else:
        print(temp, 'C: below threshold')
```

## 8. Using pandas to create a small table

Many course notebooks will use **pandas**, a Python package for working with tabular data. The table below is a tiny example dataset.

```python
import pandas as pd

data = {
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'temperature_c': [21.2, 22.5, 24.1, 23.8, 25.0],
    'rainfall_mm': [0, 4, 12, 3, 0]
}

df = pd.DataFrame(data)
df
```

## 9. View and summarize a table

These commands are commonly used at the beginning of a data-analysis workflow.

```python
print('First rows of the table:')
display(df.head())

print('Summary:')
display(df.describe())
```

## 10. Add a simple label

Sometimes we create a new column based on a rule. This is not machine learning yet, but it shows how data can be organized before analysis.

```python
df['rain_status'] = df['rainfall_mm'].apply(lambda x: 'Rain day' if x > 0 else 'No rain')
df
```

## 11. Make a simple plot

Visualizing data helps us understand patterns before using AI or machine learning methods.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.plot(df['day'], df['temperature_c'], marker='o')
plt.xlabel('Day')
plt.ylabel('Temperature (C)')
plt.title('Sample Temperature Time Series')
plt.grid(True)
plt.show()
```
